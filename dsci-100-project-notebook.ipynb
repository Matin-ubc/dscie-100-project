#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
set.seed(23145)  # setting the seed so we don't get different results each time

basketball_split <- initial_split(basketball_tidy, prop = 0.75, strata = made_top4) # we perform the train/test split 
basketball_training <- training(basketball_split) # training dataset
basketball_testing <- testing(basketball_split)   # testing dataset

basketball_recipe <- recipe(made_top4 ~ ., data = basketball_training) |>   # creating the recipe and choosing the predictors
    step_impute_mean(all_predictors()) |> # choosing the mean value for the missing value (acquired from GenAI)
    step_scale(all_predictors()) |>       # standardizing the data
    step_center(all_predictors())

basketball_tune <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>    # model specifications, we set the neighbors to tune() to find the best value of K
    set_engine("kknn") |>         # setting the engine to 'kknn' (nearest-neighbor)
    set_mode("classification")    # setting the mode to classification

k_vals <- tibble(neighbors = seq(from = 1, to = 100, by = 2))  # creating different possible values for k to test them

basketball_vfold <- vfold_cv(basketball_training, v = 10, strata = made_top4) # creating 10 fold cross-validations

basketball_possible_ks <- workflow() |>       # fitting the model using workflow and the recipe and model specs from earlier
    add_recipe(basketball_recipe) |>
    add_model(basketball_tune) |>
    tune_grid(resamples = basketball_vfold, grid = k_vals) |>  # trying different values of K on 10 cross-validations to find the best one
    collect_metrics()    # creating the table of results for each value of k
  
accuracies <- basketball_possible_ks |>   # finding the best K through the highest mean value (accuracy)
    filter(.metric == "accuracy")

best_k <- accuracies |>
    arrange(-mean) |>
    slice(1) |>
    pull(neighbors)

best_k   # so the best value for K from 1 to 100 is 9.
```

```r
# now we train our model using K = 9

best_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k) |>    # model specifications with the best K
    set_engine("kknn") |>         # setting the engine to 'kknn' (nearest-neighbor)
    set_mode("classification")    # setting the mode to classification

basketball_fit <- workflow() |>   # fitting the model on training dataset using K = 9 using the recipe and the best_spec
    add_recipe(basketball_recipe) |>
    add_model(best_spec) |>
    fit(data = basketball_training)

basketball_fit
```

```r
# Now we predict the made_top4 on testing set

basketball_test_preds <- predict(basketball_fit, basketball_testing) |>  # predicting the made_top4 factor using our final model
    bind_cols(basketball_testing)  # adding the predictions to the testing dataset

tail(basketball_test_preds)
```

```r
basketball_test_preds |> pull(made_top4) |> levels()  # checking the levels to choose which one to use for calculating precision (in this case 'second')
```

```r
top4_yes_precision <- basketball_test_preds |>
        precision(truth = made_top4, estimate = .pred_class, event_level = "second") # calculating the precision based on 'yes' in made_top4 

top4_yes_precision  # the table shows that the model's precision is 71.4% which is reasonable, but definitely not good.
```
